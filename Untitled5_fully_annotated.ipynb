### Model-specific transformation pipeline
Creates copies of the training/test features, log-transforms income, and standardizes features. Importantly, the scaler is fitted only on training data and then applied to test data, preventing **data leakage**. The transformed dataset is then used to train logistic regression.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import numpy as np

# Build logreg-specific feature sets (don't touch X_train/X_test, keep those raw for tree models)
X_train_lr = X_train.copy()
X_test_lr = X_test.copy()

X_train_lr['MonthlyIncome'] = np.log1p(X_train_lr['MonthlyIncome'])
X_test_lr['MonthlyIncome'] = np.log1p(X_test_lr['MonthlyIncome'])

scaler = StandardScaler()
X_train_lr_scaled = scaler.fit_transform(X_train_lr)
X_test_lr_scaled = scaler.transform(X_test_lr)   # fit on train only, transform test — avoids leakage

logreg = LogisticRegression(penalty='l2', C=1.0, max_iter=1000, random_state=42)
logreg.fit(X_train_lr_scaled, y_train)
```

### Train a tree-based model

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    class_weight='balanced',   # helps with the class imbalance
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)
```

### Train an imbalance-aware boosting model
Computes the class-imbalance ratio and trains XGBoost with that weighting. This is ML modeling rather than data engineering, but it consumes the curated dataset produced by the earlier pipeline.


```python
from xgboost import XGBClassifier

# scale_pos_weight handles imbalance for XGBoost specifically
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    scale_pos_weight=scale_pos_weight,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)
xgb.fit(X_train, y_train)
```

### Standardized model evaluation
Runs multiple models against the held-out test data and stores ROC-AUC and PR-AUC results in a comparison table. This creates a consistent evaluation dataset for model selection.


### Model Evaluation — Ranking Performance
This cell compares all trained classifiers on the same held-out test data. **ROC-AUC** measures how well a model ranks defaulting borrowers above non-defaulting borrowers across all thresholds. **PR-AUC (Average Precision)** is especially useful when defaults are the minority class because it emphasizes precision–recall performance on the positive/default class. Using probabilities rather than only hard predictions gives a threshold-independent comparison.


```python
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report

models = {
    'Logistic Regression': (logreg, X_test_lr_scaled),
    'Random Forest': (rf, X_test),
    'XGBoost': (xgb, X_test)
}

results = []
for name, (model, X_test_version) in models.items():
    probs = model.predict_proba(X_test_version)[:, 1]
    preds = model.predict(X_test_version)
    auc = roc_auc_score(y_test, probs)
    pr_auc = average_precision_score(y_test, probs)
    results.append({'Model': name, 'ROC-AUC': auc, 'PR-AUC': pr_auc})

pd.DataFrame(results).sort_values('ROC-AUC', ascending=False)
```

### Quantify calibration quality
Calculates Brier score and Expected Calibration Error (ECE), turning calibration quality into measurable validation metrics.


### Quantifying Calibration — Brier Score and ECE
The **Brier score** measures the mean squared error between predicted probabilities and actual binary outcomes; lower is better. **Expected Calibration Error (ECE)** groups predictions into probability bins and measures the weighted gap between predicted risk and observed default frequency. Together these metrics quantify whether the model’s probability estimates can be trusted as risk estimates.


```python
from sklearn.metrics import brier_score_loss
import numpy as np

brier = brier_score_loss(y_test, xgb_probs)
print(f"Brier score: {brier:.4f}")

# Expected Calibration Error (manual calculation, sklearn doesn't have this built in)
def expected_calibration_error(y_true, y_prob, n_bins=10):
    bin_edges = np.linspace(0, 1, n_bins + 1)
    bin_indices = np.digitize(y_prob, bin_edges[1:-1])
    ece = 0
    for b in range(n_bins):
        mask = bin_indices == b
        if mask.sum() > 0:
            bin_acc = y_true[mask].mean()
            bin_conf = y_prob[mask].mean()
            bin_weight = mask.sum() / len(y_true)
            ece += bin_weight * abs(bin_acc - bin_conf)
    return ece

ece = expected_calibration_error(y_test.values, xgb_probs)
print(f"ECE: {ece:.4f}")
```

### Create separate calibration and final-test datasets
Splits the existing test set into calibration and final evaluation subsets. This maintains separation between the data used to adjust probabilities and the data used for unbiased final assessment.


```python
from sklearn.model_selection import train_test_split

X_calib, X_test_final, y_calib, y_test_final = train_test_split(
    X_test, y_test, test_size=0.5, stratify=y_test, random_state=42
)

print(X_calib.shape, X_test_final.shape)
```

### Build calibrated probability models
Applies Platt scaling and isotonic regression to the already-trained XGBoost model using the calibration dataset. This is a post-model transformation that improves probability reliability without retraining the underlying classifier.


### Probability Calibration Models
This cell fits two post-processing calibration methods to the already-trained XGBoost model. **Platt scaling** learns a sigmoid mapping from raw scores to calibrated probabilities, while **isotonic regression** learns a flexible monotonic mapping. `FrozenEstimator` keeps the underlying XGBoost model fixed so only its probability mapping is recalibrated.


```python
from sklearn.frozen import FrozenEstimator
from sklearn.calibration import CalibratedClassifierCV

platt_model = CalibratedClassifierCV(FrozenEstimator(xgb), method='sigmoid')
platt_model.fit(X_calib, y_calib)

isotonic_model = CalibratedClassifierCV(FrozenEstimator(xgb), method='isotonic')
isotonic_model.fit(X_calib, y_calib)
```

### Produce comparable risk-score datasets
Generates uncalibrated, Platt-calibrated, and isotonic-calibrated probabilities for the same final test records so the methods can be compared consistently.


### Generate Comparable Calibrated Probabilities
The original XGBoost, Platt-scaled model, and isotonic-calibrated model all score the same final test set. Keeping the evaluation dataset identical makes the subsequent calibration comparison fair.


```python
xgb_probs_final = xgb.predict_proba(X_test_final)[:, 1]
platt_probs_final = platt_model.predict_proba(X_test_final)[:, 1]
isotonic_probs_final = isotonic_model.predict_proba(X_test_final)[:, 1]
```

### Compare calibration transformations visually
Plots reliability curves for the raw and calibrated probability outputs against the ideal calibration line.


### Before-vs-After Calibration Evaluation
These reliability curves visually compare raw XGBoost probabilities with Platt and isotonic calibration. The method whose curve stays closest to the perfect-calibration diagonal provides probabilities that most closely match observed default rates.


```python
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt

plt.figure(figsize=(7,7))

for probs, label in [
    (xgb_probs_final, 'Uncalibrated XGBoost'),
    (platt_probs_final, 'Platt scaling'),
    (isotonic_probs_final, 'Isotonic regression')
]:
    prob_true, prob_pred = calibration_curve(y_test_final, probs, n_bins=10, strategy='quantile')
    plt.plot(prob_pred, prob_true, marker='o', label=label)

plt.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Perfectly calibrated')
plt.xlabel('Mean predicted probability')
plt.ylabel('Fraction of actual positives')
plt.title('Reliability Diagram — Before vs After Calibration')
plt.legend()
plt.show()
```

### Numerical Calibration Comparison
This cell compares the three probability outputs using **Brier score** and **ECE** on the same held-out data. Lower values indicate better probability calibration. This complements the reliability diagram with numerical evidence for selecting a calibration method.


```python
from sklearn.metrics import brier_score_loss

for probs, label in [
    (xgb_probs_final, 'Uncalibrated XGBoost'),
    (platt_probs_final, 'Platt scaling'),
    (isotonic_probs_final, 'Isotonic regression')
]:
    brier = brier_score_loss(y_test_final, probs)
    ece = expected_calibration_error(y_test_final.values, probs)
    print(f"{label:25s} Brier: {brier:.4f}   ECE: {ece:.4f}")
```
